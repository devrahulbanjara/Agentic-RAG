```python
from google import genai
from google.genai import types
from pydantic import BaseModel, Field

client = genai.Client(api_key="")


class AIExplanation(BaseModel):
    core_concept: str = Field(description="The fundamental definition of AI.")
    key_steps: list[str] = Field(
        description="Sequential steps on how it processes data."
    )
    cat_fact: str = Field(
        description="A quirky closing comment or thought, speaking as a cat."
    )


# image = Image.open("/path/to/organ.png")

response = client.models.generate_content(
    model="gemini-3.1-flash-lite",
    contents=["Explain how AI works"],
    config=types.GenerateContentConfig(
        temperature=0.1,
        thinking_config=types.ThinkingConfig(thinking_level="minimal"),
        system_instruction="You are a cat. Your name is Neko.",
        response_mime_type="application/json",
        response_schema=AIExplanation,
    ),
)

structured_data: AIExplanation = response.parsed

print(f"Concept: {structured_data.core_concept}")
print(f"Steps: {structured_data.key_steps}")
print(f"Neko says: {structured_data.cat_fact}")
```
